```python
#now here we are going to create a dictionary

fruit_protein = {
    "Avocado": 2.0,       # grams of protein
    "Guava": 2.6,
    "Blackberries": 2.0,
    "Oranges": 0.9,
    "Banana": 1.1,
    "Apples": 0.3,
    "Kiwi": 1.1,
    "Pomegranate": 1.7,
    "Mango": 0.8,
    "Cherries": 1.0
}
```

```python
s2 = pd.Series(fruit_protein , name = "protien")
s2
```

```python
#conditional selection

s2[s2>1]
```

```python
# Logical Operators: and, or, not

s2[(s2>0.5) | (s2<=2)]
```

```python
# not operation:

s2[~(s2>1)]
```

```python
#update series

s2["Mango"] = 2.8
```

```python
# Brodcasting

df["Salary"] = df["Salary"] + 5000
```

```python
df["Promoted Salary"] = df["Salary"] * 10
```

```python
# Data Cleaning

df.isnull().sum()
```

```python
df.dropna(how = "any") #any row that had any null value
```

```python
df.dropna(how = "all") #if all values in a row are null then we drop that row
```

```python
df
```

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].mean())
```

```python
df["Age"].fillna(method = "ffill")
```

```python
df["Age"].fillna(method = "bfill")
```

```python
df["Name"] = df["Name"].replace("Charlie", "Rose")
```

```python
df
```

```python
# Duplicates

df_dup = df[df.duplicated(keep = "last")]
df_dup
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
# invalid values:
# Lambda -> python


df["Promoted Salary"] = df["Promoted Salary"].apply(lambda x: x/10 if x > 650000 else x)
```

```python
df
```

```python
# apply and lambda

def multiplying_age(x):
    return x*2
df["Age"] = df["Age"].apply(multiplying_age)
```

```python
df
```

```python
df["Age"] = df["Age"].apply(lambda x: x/2)
```

```python
df
```

```python
# Joins and merges

department_info = {
    "Dept" : ["HR", "IT", "Finance"],
    "Location" : ["New York", "San Francisco", "Chicago"],
    "Manager" : ["Laura", "Steve", "Nina"]
}

df2 = pd.DataFrame(department_info)
df2
```

```python
# Concat

pd.concat([df, df2])
```

```python
pd.concat([df, df2], axis = 1)
```

```python
pd.merge(df, df2, on = "Dept")
```
